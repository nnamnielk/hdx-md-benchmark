```python
from IPython.display import display, HTML
display(HTML("<style>.container { width:90% !important; }</style>"))

import sys, os, re, glob
from math import ceil
import numpy as np
import pandas as pd
import seaborn as sb
from scipy.spatial.transform import Rotation as R
from scipy import stats
import mdtraj as md
import tables as tb
import nglview as nv
# %matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from mpl_toolkits.mplot3d import Axes3D
# import matplotlib.matlab.hist as hist

upside_home = os.environ["UPSIDE_HOME"]
upside_utils_dir = os.path.join(upside_home, "py")
sys.path.append(upside_utils_dir)
import mdtraj_upside as mu

mpl.rcParams['font.size'] = 18
mpl.rcParams['axes.labelsize'] = 'medium'
mpl.rcParams['ytick.labelsize'] = 'medium'
mpl.rcParams['xtick.labelsize'] = 'medium'
mpl.rcParams['axes.linewidth'] = 2
mpl.rcParams['xtick.major.width'] = 2
mpl.rcParams['xtick.minor.width'] = 2
mpl.rcParams['ytick.major.width'] = 2
mpl.rcParams['ytick.minor.width'] = 2
mpl.rcParams['font.sans-serif'] = "Arial"
mpl.rcParams['font.family'] = "sans-serif"
mpl.rcParams['svg.fonttype'] = 'none'

base_dir = os.path.join(os.getcwd(), os.pardir, "simulations", "nug2b") + "/"
up_indir = base_dir + "inputs/hdx/"
pdb_id = "nug2b"
n_rep = 20

output_dir = base_dir + "outputs/"

init_fn = up_indir + "{}.initial.npy".format(pdb_id)
fasta_fn = up_indir + "{}.fasta".format(pdb_id)
chain_breaks_fn = up_indir + "{}.chain_breaks".format(pdb_id)
if not os.path.exists(chain_breaks_fn):
    chain_breaks_fn = ""

# Each .run.N.up is a fixed-temperature REMD *slot*, not one walker: replica_index
# changes on ~87% of consecutive frames, so neighbouring frames are unrelated
# structures. load_upside_rep demultiplexes to follow a single replica.
run_fns = sorted(glob.glob(output_dir + "remd/{}.run.*.up".format(pdb_id)),
                 key=lambda f: int(re.search(r"run\.(\d+)\.up$", f).group(1)))
rep_select = 0
traj = mu.load_upside_rep(run_fns, rep_select)

# Remove global tumbling/translation so the animation shows conformational change.
ca_idx = traj.top.select("name CA")
traj.superpose(traj, 0, atom_indices=ca_idx)

acid_sele = traj.top.select("(resname ASP or resname GLU) and name CA")
base_sele = traj.top.select("(resname LYS or resname ARG) and name CA")
print(traj)
```

```python
from ipywidgets import Play, IntSlider, HBox, VBox, Label, link

view = nv.NGLWidget()
# default_representation=False suppresses NGL's own colour-by-chain backbone at
# load time. clear_representations() cannot do this reliably -- it races against
# the representation that add_trajectory queues, which is why both a coloured and
# a grey backbone showed up.
view.add_trajectory(traj, default_representation=False)
view.add_cartoon(color='#888888')
view.add_spacefill(selection=acid_sele, color='#ff0000')
view.add_spacefill(selection=base_sele, color='#0000ff')

view.camera = 'orthographic'
view.parameters = {
    "clipNear": -100, "clipFar": 1000,
    "fogNear": 0, "fogFar": 1000,
}

# nglview's built-in view._iplayer wires its Play/slider maxes with jslink, which
# does not propagate in VS Code -- the widgets keep max=0, so only a few frames
# play. Build the player here with an explicit max and kernel-side link instead.
n_frames = traj.n_frames
view.max_frame = n_frames - 1

player = Play(value=0, min=0, max=n_frames - 1, step=1, interval=100)
slider = IntSlider(value=0, min=0, max=n_frames - 1, continuous_update=True,
                   readout=False, layout={'width': '400px'})
readout = Label(value="frame 0 / {}".format(n_frames - 1))

link((player, 'value'), (slider, 'value'))
link((slider, 'value'), (view, 'frame'))
slider.observe(lambda c: setattr(readout, 'value',
                                 "frame {} / {}".format(c['new'], n_frames - 1)),
               names='value')

VBox([view, HBox([player, slider, readout])])
```

```python
# Compute DSSP secondary structure per frame using MDTraj
dssp = md.compute_dssp(traj, simplified=True)

# Color scheme for secondary structure:
# H = Helix (Red), E = Sheet (Blue), C = Coil/Loop (Gray)
dssp_colors = {
    "H": "#ff2255",
    "E": "#0099ff",
    "C": "#aaaaaa"
}

def apply_dssp_coloring(view_obj, frame_idx):
    view_obj.clear_representations()
    current_dssp = dssp[frame_idx]
    for ss_code, color in dssp_colors.items():
        res_seqs = [str(traj.top.residue(i).resSeq) for i, ss in enumerate(current_dssp) if ss == ss_code]
        if res_seqs:
            selection_str = " or ".join(res_seqs)
            view_obj.add_cartoon(selection=selection_str, color=color)

# Create NGL view for DSSP trajectory player
view_dssp = nv.NGLWidget()
view_dssp.add_trajectory(traj, default_representation=False)

# Initial coloring for frame 0
apply_dssp_coloring(view_dssp, 0)

view_dssp.camera = "orthographic"
view_dssp.parameters = {
    "clipNear": -100, "clipFar": 1000,
    "fogNear": 0, "fogFar": 1000,
}

n_frames = traj.n_frames
view_dssp.max_frame = n_frames - 1

# Widgets setup (Play button & Slider with dynamic DSSP updates)
player_dssp = Play(value=0, min=0, max=n_frames - 1, step=1, interval=100)
slider_dssp = IntSlider(value=0, min=0, max=n_frames - 1, continuous_update=True,
                        readout=False, layout={"width": "400px"})
readout_dssp = Label(value="frame 0 / {}".format(n_frames - 1))

link((player_dssp, "value"), (slider_dssp, "value"))
link((slider_dssp, "value"), (view_dssp, "frame"))

def on_dssp_frame_change(change):
    frame = change["new"]
    readout_dssp.value = "frame {} / {}".format(frame, n_frames - 1)
    apply_dssp_coloring(view_dssp, frame)

slider_dssp.observe(on_dssp_frame_change, names="value")

VBox([view_dssp, HBox([player_dssp, slider_dssp, readout_dssp])])
```
